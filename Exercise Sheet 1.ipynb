Problem 5: Write code to display only specific items from a list/array that satisfy the following conditions:
• The number must be divisible by five.
• If the number is greater than or equal to 150, then skip it.
• If the number is greater than 500, then stop the loop.
Example input:
• Python: [12, 75, 150, 182, 145, 525, 50]

if number > 500
    break
else if number >= 150
    continue
   else if number mod 5 = 0
        print number
end if

```python
numbers = [12, 75, 150, 182, 145, 525, 50]

for number in numbers:
    if number > 500:
        break
    elif number >= 150:
        continue
    elif number % 5 == 0:
        print(number)
    else:
        continue
```
