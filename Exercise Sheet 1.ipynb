Problem 5: Write code to display only specific items from a list/array that satisfy the following conditions:
• The number must be divisible by five.
• If the number is greater than or equal to 150, then skip it.
• If the number is greater than 500, then stop the loop.
Example input:
• Python: [12, 75, 150, 182, 145, 525, 50]

if number > 500
    break
else if number >= 150
    continue
   else if number mod 5 = 0
        print number
end if

```python
numbers = [12, 75, 150, 182, 145, 525, 50]

for number in numbers:
    if number > 500:
        break
    elif number >= 150:
        continue
    elif number % 5 == 0:
        print(number)
    else:
        continue
```

```python
#another way

x = [12, 75, 150, 182, 145, 525, 50]

#print(type(x))

for i in x:
    if i % 5 == 0 and i < 150:
        print(i)
    elif i > 500:
        break
```

```python
start = 0
stop = 10
step = 1

list(range(start, stop, step))

n = 25
m= 50

for i in range (n, m + 1, 1):
    # checking if i is prime by going from i down to 1
    j = i - 1
    while j > 1:
        if i % j == 0:
            break # when i is not prime
        j = j - 1
    if j == 1:
        print(i)
```

```python
#another way for instances like 25 to 5000
def check_prime(n, m):
    for i in range (n, m + 1, 1):

        if i < 2: #checks if the number is 1 (as it is not a prime number)
            continue
        elif i % 2 == 0 and i != 2: #checks the even numbers apart from 2
            continue
        elif is_prime(i) == False: #checks only the odd numbers and 2 (makes it going only through half of the values)
            print(i)


#mathematical approach with divisible by 2 or the square root of the number

def is_prime(num):
    for j in range(3, int(num ** 0.5) + 1, 2): #checks only the odd numbers
        if num % j == 0:
            return False

n = 25
m = 50
check_prime(n, m)
```
